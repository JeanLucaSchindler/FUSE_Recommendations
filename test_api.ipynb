# Recommender system

```python
import pandas as pd
data = pd.read_csv('/home/jeanluca/code/JeanLucaSchindler/FUSE/raw_data/df_gpt_summaries.csv')
data = data.set_index(data['index'])
```

```python
data = data.drop(['Unnamed: 0', 'index'], axis=1)
```

```python
import string
from nltk import word_tokenize
from nltk.stem import WordNetLemmatizer
from nltk.corpus import stopwords

def preprocessing(sentence):

    # Remove whitespace
    sentence = sentence.strip()

    # Lowercase characters
    sentence = sentence.lower()

    #stop_word
    stop_words = set(stopwords.words('english'))
    sentence = ' '.join(word for word in sentence.split() if word not in stop_words)

    # Remove numbers
    sentence = ''.join(char for char in sentence if not char.isdigit())

    # Remove punctuation
    for punctuation in string.punctuation:
        sentence = sentence.replace(punctuation, '')

    # Tokenize
    sentence = word_tokenize(sentence)

    # Lemmatize
    verbs_lemmed = [
        WordNetLemmatizer().lemmatize(word, pos='v')
        for word in sentence
    ]
    nouns_lemmed = [
        WordNetLemmatizer().lemmatize(word, pos='n')
        for word in verbs_lemmed
    ]
    return ' '.join(word for word in nouns_lemmed)
```

```python
data['clean_text'] = data.gpt40_summary.apply(preprocessing)
```

```python
data
```

```python
from sklearn.feature_extraction.text import CountVectorizer

count_vectorizer = CountVectorizer()

count_df = count_vectorizer.fit_transform(data['clean_text'])
count_df = pd.DataFrame(count_df.toarray())
count_df
```

```python
from sklearn.decomposition import TruncatedSVD
```

```python
svd = TruncatedSVD(n_components=615)
trunc_df = svd.fit_transform(count_df)
```

```python
print(svd.explained_variance_ratio_.sum())
```

```python
trunc_df = pd.DataFrame(trunc_df, columns=[f'component_{i+1}' for i in range(trunc_df.shape[1])])
latent_df = trunc_df.set_index(data.index)
latent_df
```

```python
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np
```

```python
# Compute cosine similarity for content matrix
content_similarities = cosine_similarity(np.array(latent_df.iloc[18]).reshape(1, -1), latent_df).flatten()
```

```python
reco = pd.DataFrame(content_similarities)
reco = reco.set_index(data.index)
reco = reco.rename(columns={0:'similarity'})
```

```python
top_reco = reco.sort_values(by=['similarity'], ascending=False).head(5)
top_reco
```

```python
import random

# Create an empty list to store the random integers
random_integers = []

# Use a for loop to generate 100 random integers
for i in range(100):
    random_integers.append(random.randint(0, 1800))
```

```python
len(random_integers)
```

```python
recommendation_dict = {}
for i in random_integers:
    content_similarities = cosine_similarity(np.array(latent_df.iloc[i]).reshape(1, -1), latent_df).flatten()

    reco = pd.DataFrame(content_similarities)
    reco = reco.set_index(data.index)
    reco = reco.rename(columns={0:'similarity'})

    top_reco = reco.sort_values(by=['similarity'], ascending=False).head(5)

    recommendation_dict[top_reco.index[0]] = data.loc[top_reco.index[0]]['text'], data.loc[top_reco.index[1]]['text'], data.loc[top_reco.index[2]]['text']
```

```python
test = pd.DataFrame.from_dict(recommendation_dict)
```

```python
# Transpose the DataFrame
df_transposed = test.T

# Rename the columns to reflect Reference 1 and Reference 2
df_transposed.columns = ['Original text', 'Recommendation 1', 'Recommendation 2']

# Reset the index to get a clean DataFrame
df_transposed.reset_index(inplace=True)

# Rename the index column to something meaningful (optional)
df_transposed.rename(columns={'index': 'Reference Number'}, inplace=True)

df_transposed
```
